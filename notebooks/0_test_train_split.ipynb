```python
import os
import glob
import numpy as np
import re
import pandas as pd
import shutil
```

```python
filenames_ = glob.glob('../images/train/*.json')
```

```python
np.unique([i.split('-')[1] for i in filenames_])
```

```python
np.random.seed(2397165)
val_choices = np.random.choice(np.unique([i.split('-')[1] for i in filenames_]), size=4, replace=False)

val_choices
```

```python
data = pd.DataFrame({
    'filename': filenames_,
    'video_id': [i.split('-')[1] for i in filenames_],
    'frame': [int(re.findall('(\d+)\.json', i)[0]) for i in filenames_]})
```

```python
data['is_validation'] = data.video_id.isin(val_choices).astype('int')
```

```python
data.is_validation.mean()
```

```python
data.to_csv('package_dataset.csv', index=False)
```

```python
os.mkdir('../images/train_json')
os.mkdir('../images/val_json')
```

```python
base_path_train_json = '../images/train_json'
base_path_val_json = '../images/val_json'
base_path = '../images/train'
for n, i in data.iterrows():
    filename_ext = i['filename'].split('\\')[1]
    if i['is_validation'] == 0:
        shutil.move(i['filename'], os.path.join(base_path_train_json, filename_ext))
    else:
        shutil.move(i['filename'], os.path.join(base_path_val_json, filename_ext))
```
